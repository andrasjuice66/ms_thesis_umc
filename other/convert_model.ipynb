```python
import os
import sys
import numpy as np
import tensorflow as tf

# Add SynthSeg to path
synthseg_path = "C:/Projects/thesis_project/OtherRepos/SynthSeg"
if synthseg_path not in sys.path:
    sys.path.append(synthseg_path)

# Import SynthSeg modules
from ext.neuron import models as nrn_models
from ext.lab2im import utils

def load_synthseg_model(model_path, robust=False):
    """
    Load SynthSeg model by building architecture and loading weights
    
    Args:
        model_path: Path to the .h5 weights file
        robust: Whether to use the robust model architecture
    """
    # Standard SynthSeg segmentation labels (you may need to adjust these)
    labels_segmentation = [0, 2, 3, 4, 5, 7, 8, 10, 11, 12, 13, 14, 15, 16, 17, 18, 26, 28, 
                          41, 42, 43, 44, 46, 47, 49, 50, 51, 52, 53, 54, 58, 60]
    
    n_labels_seg = len(labels_segmentation)
    
    if robust:
        # For robust models like synthseg_robust_2.0.h5
        # This uses a more complex architecture with denoising
        labels_denoiser = [0, 1]  # Simplified for this example
        n_groups = len(labels_denoiser)
        
        # Build first UNet for denoising
        net = nrn_models.unet(input_shape=[None, None, None, 1],
                              nb_labels=n_groups,
                              nb_levels=5,
                              nb_conv_per_level=2,
                              conv_size=3,
                              nb_features=24,
                              feat_mult=2,
                              activation='elu',
                              batch_norm=-1,
                              name='unet')
        
        # Build the full robust architecture (simplified version)
        # This would need the full robust architecture from predict_synthseg.py
        # For now, let's use the standard architecture
        net = nrn_models.unet(input_shape=[None, None, None, 1],
                              nb_labels=n_labels_seg,
                              nb_levels=5,
                              nb_conv_per_level=2,
                              conv_size=3,
                              nb_features=24,
                              feat_mult=2,
                              activation='elu',
                              batch_norm=-1,
                              name='unet')
    else:
        # Standard SynthSeg architecture
        net = nrn_models.unet(input_shape=[None, None, None, 1],
                              nb_labels=n_labels_seg,
                              nb_levels=5,
                              nb_conv_per_level=2,
                              conv_size=3,
                              nb_features=24,
                              feat_mult=2,
                              activation='elu',
                              batch_norm=-1,
                              name='unet')
    
    # Load weights
    net.load_weights(model_path, by_name=True)
    
    return net

# Load the standard SynthSeg model
model_path = "C:/Projects/thesis_project/OtherRepos/SynthSeg/models/synthseg_2.0.h5"
loaded_model = load_synthseg_model(model_path, robust=False)

print("Model loaded successfully!")
print(f"Model summary:")
loaded_model.summary()
```
